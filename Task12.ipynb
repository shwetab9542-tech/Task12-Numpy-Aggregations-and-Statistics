```python
import numpy as np
from sklearn.datasets import load_iris
```

```python
#Load the Iris Dataset
iris = load_iris()

X = iris.data
feature_names = iris.feature_names

print("Dataset Shape:", X.shape)
print("Features:", feature_names)
```

```python
print("First 10 rows of Iris Dataset:")
print(X[:10])
```

```python
#Calculate Sum
total = np.sum(X, axis=0)

print("Sum of each feature:")
print(total)
```

```python
#Calculate Mean
mean = np.mean(X, axis=0)

print("Mean of each feature:")
print(mean)
```

```python
#Calculate Median
median = np.median(X, axis=0)

print("Median of each feature:")
print(median)
```

```python
#Calculate Minimum and Maximum
minimum = np.min(X, axis=0)
maximum = np.max(X, axis=0)

print("Minimum values:")
print(minimum)

print("\nMaximum values:")
print(maximum)
```

```python
#Calculate Standard Deviation
std = np.std(X, axis=0)

print("Standard deviation:")
print(std)
```

```python
#Calculate Variance
variance = np.var(X, axis=0)

print("Variance:")
print(variance)
```

```python
#Compare Using axis=0 
print("Mean using axis=0:")
print(np.mean(X, axis=0))
```

```python
#Axis = 1
print("Mean of first 10 samples using axis=1:")
print(np.mean(X[:10], axis=1))
```

```python
#Create a Statistical Summary
print("Statistical Summary")

print("Mean:", np.mean(X, axis=0))
print("Median:", np.median(X, axis=0))
print("Minimum:", np.min(X, axis=0))
print("Maximum:", np.max(X, axis=0))
print("Standard Deviation:", np.std(X, axis=0))
print("Variance:", np.var(X, axis=0))
print("Sum:", np.sum(X, axis=0))
```
